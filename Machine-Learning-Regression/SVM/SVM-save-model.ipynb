```python
#split into training set and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(independent_var, denpendent_var, test_size = 0.3, random_state = 0)
```

```python
# create the model-svm-linear
from sklearn.svm import SVR
regressor = SVR(kernel="linear")
regressor.fit(X_train, y_train)
```

```python
# Viewing the b1 and bo value
weight=regressor.coef_
print("Weight of the model={}".format(weight))
bais=regressor.intercept_
print("Intercept of the model={}".format(bais))
```

```python
# predict model with test data
y_pred=regressor.predict(X_test)
```

```python
# evaluate the model and determine accuracy
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
```

```python
# Scaling features (crucial for SVM)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# 3. Define Parameter Grid
kernels = ['linear', 'rbf','poly']
Cs = [0.1, 1]
epsilons = [0.1, 0.2,0.5]
gammas = ['scale','auto']
degree=[3,5]
coef=[1,2]

# 4. Manual Loop and Storage
results = []

for kernel in kernels:
    for c in Cs:
        for eps in epsilons:
            for gs in gammas:
                for dg in degree:
                    for cf in coef:
                        # Initialize and train SVR
                        model = SVR(kernel=kernel, C=c, gamma=gs, degree=dg, epsilon=eps, coef0=cf)
                        model.fit(X_train_scaled, y_train)
                        
                        # Predict and calculate R^2 score
                        predictions = model.predict(X_test_scaled)
                        r2 = r2_score(y_test, predictions)
                        
                        # Store results
                        results.append({
                            'kernel': kernel,
                            'C': c,
                            'epsilon': eps,
                            'r2_score': r2
                        })
```
