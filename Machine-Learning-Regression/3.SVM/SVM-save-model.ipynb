```python
# Importing the libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import warnings
from sklearn.preprocessing import StandardScaler
warnings.filterwarnings('ignore')
```

```python
data_set = pd.read_csv('50_Startups.csv')
```

```python
data_set
```

```python
data_set.columns
```

```python
data_set_m = pd.get_dummies(data_set, drop_first=True)
```

```python
data_set_m.columns
```

```python
independent_var = data_set_m[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
denpendent_var = data_set_m[[ 'Profit']]
```

```python
independent_var
```

```python
denpendent_var
```

```python
#split into training set and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(independent_var, denpendent_var, test_size = 0.3, random_state = 0)
```

```python
X_train.shape
```

```python
 X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
# create the model-svm-linear
from sklearn.svm import SVR
regressor = SVR(kernel="linear",C=1000)
regressor.fit(X_train, y_train)
```

```python
# predict model with test data
y_pred=regressor.predict(X_test)
```

```python
# evaluate the model and determine accuracy
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
```

```python
r_score
```

```python
# accuracy = 89%
```

```python
# Scaling features (crucial for SVM)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
regressor = SVR(kernel="linear",)
regressor.fit(X_train_scaled, y_train)
# predict model with test data
y_pred=regressor.predict(X_test_scaled)
```

```python
r_score=r2_score(y_test,y_pred)
```
