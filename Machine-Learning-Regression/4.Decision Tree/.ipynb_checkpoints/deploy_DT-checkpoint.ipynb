```python
import joblib
import numpy as np

# Load model and scaler
model = joblib.load("models/decision_tree_v1.joblib")
```

```python
# Example: Suppose your model expects 5 features

f1 = float(input("Enter R&D Spend: "))
f2 = float(input("Enter Administration: "))
f3 = float(input("Enter Marketing Spend: "))
f4 = float(input("Enter State_Florida 0 or 1: "))
f5 = float(input("Enter State_New York 0 or 1: "))

# Combine into array
user_data = np.array([[f1, f2, f3,f4,f5]])
prediction = model.predict(user_data)
print("Predicted Value:", prediction[0])
```
